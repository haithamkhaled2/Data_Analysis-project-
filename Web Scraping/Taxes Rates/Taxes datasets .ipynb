```python
import pandas as pd 
from bs4 import BeautifulSoup
import requests
```

```python
url='https://taxfoundation.org/data/all/federal/2025-tax-brackets/'
page=requests.get(url)
page
```

```python
soup=BeautifulSoup(page.content,'html.parser')
soup
table=soup.find_all('table')[0]  
table
```

```python
table_indexing = table.find_all('th')
table_indexing
table_indexing_title = [title.text.strip() for title in table_indexing]
table_indexing_title 
```

```python
df=pd.DataFrame(columns=table_indexing_title)
df
```

```python
column_rows = table.find_all('tr')
for row in column_rows[1:]:
    row_contents= row.find_all('td')
    data_row = [data.text.strip() for data in row_contents]
    df.loc[len(df)]= data_row
df
```
