Libraries

```python
from bs4 import BeautifulSoup
from urllib.request import urlopen
import requests
import pandas as pd
```

Variables

```python
NumPage = 0
job_title = []
company_name = []
location_name = []
status = []
skills = []
links = []
```

Processing & Scarping

```python

while NumPage < 10:
    url = f"https://wuzzuf.net/search/jobs/?a=hpb%7Cspbg&q=python&start={NumPage}"
    try:
        # client = urlopen(url)
        # html = client.read()
        # client.close()
        responds = requests.get(url)
        html = responds.content
    except:
        break

    soup = BeautifulSoup(html, 'html.parser')

    job_titles = soup.find_all('h2', {'class': 'css-m604qf'})
    company_names = soup.find_all('a', {'class': 'css-17s97q8'})
    location_names = soup.find_all('span', {'class': 'css-5wys0k'})
    job_statuses = soup.find_all('div', {'class': 'css-1lh32fc'})
    job_skills = soup.find_all('div', {'class': 'css-y4udm8'})

    min_length = min(len(job_titles), len(company_names), len(location_names), len(job_statuses), len(job_skills))
    
    for i in range(min_length):
        job_title.append(job_titles[i].text.strip())
        links.append(job_titles[i].find('a').attrs['href'])
        company_name.append(company_names[i].text.strip())
        location_name.append(location_names[i].text.strip())
        status.append(job_statuses[i].text.strip())
        skills.append(job_skills[i].text.strip())

    NumPage += 1



```

Storing

```python
job_detail = {
    'Job Title': job_title,
    'Company Name': company_name,
    'Location': location_name,
    'Status': status,
    'Skills Requirements': skills,
    'Links': links
}
```

Convert to Excel File

```python
df = pd.DataFrame(job_detail)
df
# df.to_excel('Wuzzuf Python Jobs Opportunity.xlsx' , index = False)
```
