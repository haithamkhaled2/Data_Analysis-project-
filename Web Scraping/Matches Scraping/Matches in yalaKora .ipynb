```python
from bs4 import BeautifulSoup
import requests 
import csv
import lxml
import re 
import pandas as pd 
```

```python
url = "https://www.yallakora.com/match-center?date=7/5/2025"
page=requests.get(url)

src= page.content
soup = BeautifulSoup(src, 'lxml')

championships = soup.find_all('div',{'class': 'matchCard'})

matches_detail = []

for championship in championships:
    champion = championship.find('h2').text.strip()
    matches_list = championship.find('div', {'class': 'ul'})
    if not matches_list:
        continue
    matches = matches_list.find_all('div', {'class': 'item finish liItem'})

    for match in matches:
        team1 = match.find('div', {'class': 'teamA'}).text.strip()
        team2 = match.find('div', {'class': 'teamB'}).text.strip()
        score = match.find('div', {'class': 'MResult'}).find_all('span', {'class': 'score'})
        result = f"'{score[0].text.strip()}' - '{score[1].text.strip()}'"
        time = match.find('div', {'class': 'MResult'}).find('span', {'class': 'time'}).text.strip()
        matches_detail.append({
            'champion Type': champion,
            'teamA': team1,
            'teamB': team2,
            'result': result,
            'time': time
        })

# Then convert to DataFrame and save
df = pd.DataFrame(matches_detail)
df.to_excel(r"C:\Users\haith\PyFolder\Matches_Check.xlsx", index=False)
df
```
